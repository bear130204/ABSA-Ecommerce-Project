```python
import pandas as pd
import numpy as np
import torch

from datasets import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

from sklearn.preprocessing import LabelEncoder

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("CUDA chưa bật. Nếu train bằng CPU sẽ rất chậm.")
```

```python
from pathlib import Path

BASE_DIR = Path.cwd()

# nếu notebook nằm trong notebooks/ thì đi lùi 1 cấp về project root
if BASE_DIR.name == "notebooks":
    PROJECT_ROOT = BASE_DIR.parent
else:
    PROJECT_ROOT = BASE_DIR

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
MODEL_DIR = PROJECT_ROOT / "models_saved" / "xlmr_aspect_model"

MODEL_DIR.mkdir(parents=True, exist_ok=True)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("PROCESSED_DIR:", PROCESSED_DIR)
print("MODEL_DIR:", MODEL_DIR)

train_df = pd.read_csv(PROCESSED_DIR / "train.csv")
valid_df = pd.read_csv(PROCESSED_DIR / "valid.csv")
test_df = pd.read_csv(PROCESSED_DIR / "test.csv")

print("Train:", train_df.shape)
print("Valid:", valid_df.shape)
print("Test :", test_df.shape)

train_df.head()
```

```python
train_df = train_df[train_df["aspect"] != "translation"].reset_index(drop=True)
valid_df = valid_df[valid_df["aspect"] != "translation"].reset_index(drop=True)
test_df = test_df[test_df["aspect"] != "translation"].reset_index(drop=True)

print("Train:", train_df.shape)
print("Valid:", valid_df.shape)
print("Test :", test_df.shape)

print(train_df["aspect"].value_counts())
```

```python
label_encoder = LabelEncoder()

train_df["label"] = label_encoder.fit_transform(train_df["aspect"])
valid_df["label"] = label_encoder.transform(valid_df["aspect"])
test_df["label"] = label_encoder.transform(test_df["aspect"])

num_labels = len(label_encoder.classes_)

print("Aspect classes:")
print(label_encoder.classes_)

print("Num labels:", num_labels)
```

```python
train_df["text"] = train_df["review"].astype(str)
valid_df["text"] = valid_df["review"].astype(str)
test_df["text"] = test_df["review"].astype(str)

train_df[["text", "aspect", "label"]].head()
```

```python
train_dataset = Dataset.from_pandas(
    train_df[["text", "label"]],
    preserve_index=False
)

valid_dataset = Dataset.from_pandas(
    valid_df[["text", "label"]],
    preserve_index=False
)

test_dataset = Dataset.from_pandas(
    test_df[["text", "label"]],
    preserve_index=False
)

print(train_dataset)
print(valid_dataset)
print(test_dataset)
```

```python
MODEL_NAME = "xlm-roberta-base"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

```python
MAX_LENGTH = 128

def tokenize_function(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        padding="max_length",
        max_length=MAX_LENGTH
    )

train_dataset = train_dataset.map(
    tokenize_function,
    batched=True
)

valid_dataset = valid_dataset.map(
    tokenize_function,
    batched=True
)

test_dataset = test_dataset.map(
    tokenize_function,
    batched=True
)
```

```python
train_dataset.set_format(
    type="torch",
    columns=["input_ids", "attention_mask", "label"]
)

valid_dataset.set_format(
    type="torch",
    columns=["input_ids", "attention_mask", "label"]
)

test_dataset.set_format(
    type="torch",
    columns=["input_ids", "attention_mask", "label"]
)
```

```python
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=num_labels
)

model.to(device)

print("Model loaded on:", device)
```

```python
def compute_metrics(eval_pred):
    logits, labels = eval_pred

    preds = np.argmax(logits, axis=1)

    acc = accuracy_score(labels, preds)

    macro_f1 = f1_score(
        labels,
        preds,
        average="macro"
    )

    weighted_f1 = f1_score(
        labels,
        preds,
        average="weighted"
    )

    return {
        "accuracy": acc,
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1
    }
```

```python
training_args = TrainingArguments(
    output_dir=str(MODEL_DIR),

    eval_strategy="epoch",
    save_strategy="epoch",

    learning_rate=2e-5,

    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,

    num_train_epochs=3,

    weight_decay=0.01,

    logging_dir=str(PROJECT_ROOT / "reports" / "logs"),
    logging_steps=100,

    load_best_model_at_end=True,
    metric_for_best_model="macro_f1",
    greater_is_better=True,

    save_total_limit=2,

    fp16=torch.cuda.is_available(),

    report_to="none"
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,

    train_dataset=train_dataset,
    eval_dataset=valid_dataset,

    compute_metrics=compute_metrics
)
```

```python
trainer.train()
```

```python
valid_results = trainer.evaluate(valid_dataset)

valid_results
```

```python
test_results = trainer.evaluate(test_dataset)

test_results
```

```python
predictions = trainer.predict(test_dataset)

y_pred = np.argmax(
    predictions.predictions,
    axis=1
)

y_true = predictions.label_ids

print(
    classification_report(
        y_true,
        y_pred,
        target_names=label_encoder.classes_
    )
)
```

```python
import pickle

trainer.save_model(str(MODEL_DIR))
tokenizer.save_pretrained(str(MODEL_DIR))

with open(MODEL_DIR / "label_encoder.pkl", "wb") as f:
    pickle.dump(label_encoder, f)

print("Model saved to:", MODEL_DIR)
```

```python
def predict_aspect(text):

    inputs = tokenizer(
        text,
        truncation=True,
        padding="max_length",
        max_length=MAX_LENGTH,
        return_tensors="pt"
    )

    inputs = {
        k: v.to(device)
        for k, v in inputs.items()
    }

    model.eval()

    with torch.no_grad():
        outputs = model(**inputs)

    pred_id = torch.argmax(
        outputs.logits,
        dim=1
    ).item()

    return label_encoder.inverse_transform([pred_id])[0]
```

```python
text = "áo mặc rất thoải mái nhưng giao hàng hơi chậm"

predict_aspect(text)
```
